# Constrained residual-learning secondary model

This modified copy implements the logic in `progress.txt` while retaining the
original data preparation. The MLP predicts the primary model residual
$y^m=y-\hat y$. Its training loss is

$$
\mathcal L_{total}=\mathcal L_{predict}
+\gamma_{asset}\mathcal L_{asset}
+\gamma_{geo}\mathcal L_{geo}
+\gamma_{risk}\mathcal L_{risk}.
$$

For the constraint terms, the pre-threshold final signal
$S=\hat y+\widehat{y^m}$ is mapped to long-only differentiable weights using
`softplus` followed by cross-sectional normalization. Applying constraints to
the final signal, rather than to the residual alone, makes the penalties govern
the signal that will actually be traded.

The risk term computes $w^\top\Sigma w$ from trailing adjusted-close returns as
the sample variance of the portfolio return series. This is algebraically
equivalent to forming the covariance matrix but avoids a dense $N\times N$
matrix. To keep rolling training tractable, constraint losses are evaluated on
a configurable, evenly spaced subset of training dates.

Geographic constraints require an external `stock_geo_map.csv` with columns
`stock` and `geo`. The current A-share dataset has no such field, so
`gamma_geo=0` by default. It can be activated after supplying the mapping and
group caps.


```python
import copy
import os
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from tqdm import tqdm

CONFIG = {
    "seed": 42,
    "hidden_dim": 64,
    "epochs": 300,
    "patience": 30,
    "learning_rate": 1e-3,
    "weight_decay": 1e-4,
    "train_days": 500,
    "validation_days": 20,
    "test_days": 20,
    "step_days": 20,
    # stock_label_r10.csv is treated as a 10-trading-day forward return label.
    "label_horizon_days": 10,
    # Residual corrections smaller than epsilon are filtered at inference.
    "epsilon": 0.005,
    "gamma_asset": 10.0,
    "gamma_geo": 0.0,
    "gamma_risk": 1.0e4,
    "w_max": 0.02,
    "sigma_max": 0.02,
    "risk_lookback": 60,
    "minimum_risk_history": 30,
    # Set to None to evaluate constraints on every fitting date.
    "constraint_dates": 20,
    # Populate after supplying stock_geo_map.csv, e.g. {"CN": 0.8}.
    "geo_caps": {},
}

DATA_ROOT = "/home/yichuan/ywc/meta-labeling"
CLOSE_PATH = "/home/yichuan/ywc/preparation/BasicFactor_Close.txt"
ADJ_PATH = "/home/yichuan/ywc/preparation/BasicFactor_AdjFactor.txt"
GEO_MAP_PATH = os.path.join(DATA_ROOT, "stock_geo_map.csv")
OUTPUT_SIGNAL_PATH = os.path.join(
    DATA_ROOT, "secondary_signal_with_constraints.csv"
)
OUTPUT_DIAGNOSTICS_PATH = os.path.join(
    DATA_ROOT, "secondary_signal_with_constraints_diagnostics.csv"
)

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


set_seed(CONFIG["seed"])
print("device:", device)
```

```python
# Use explicit key checks before any positional operations.
meta_features_model = meta_features_new.copy().reset_index(drop=True)
meta_features_model["date"] = pd.to_datetime(meta_features_model["date"])

primary_frame = pri_sig[["date", "stock", "true", "pred"]].copy()
primary_frame["date"] = pd.to_datetime(primary_frame["date"])
primary_frame = primary_frame.reset_index(drop=True)

if len(meta_features_model) != len(primary_frame):
    raise ValueError("Feature rows and primary-signal rows have different lengths.")

key_match = (
    meta_features_model[["date", "stock"]].reset_index(drop=True)
    == primary_frame[["date", "stock"]].reset_index(drop=True)
).all().all()
if not key_match:
    raise ValueError("(date, stock) keys are not aligned with primary_signal.csv.")

meta_targets = pd.Series(meta_labels).reset_index(drop=True).astype(float)
primary_predictions = primary_frame["pred"].astype(float)
true_returns = primary_frame["true"].astype(float)
feature_columns = [
    col for col in meta_features_model.columns if col not in ["date", "stock"]
]

print("rows:", len(meta_features_model))
print("model features:", len(feature_columns))
print("date range:", meta_features_model["date"].min(), meta_features_model["date"].max())
```

```python
def read_wide_panel(path):
    panel = pd.read_feather(path)
    panel = dateindex(panel)
    panel.index = pd.to_datetime(panel.index)
    return panel.sort_index()


# Historical returns are used only by the risk penalty and are strictly lagged.
if CONFIG["gamma_risk"] > 0:
    close_panel = read_wide_panel(CLOSE_PATH)
    adjustment_panel = read_wide_panel(ADJ_PATH)
    adjustment_panel = adjustment_panel.reindex(
        index=close_panel.index, columns=close_panel.columns
    )
    adjusted_close = close_panel * adjustment_panel
    historical_returns = adjusted_close.pct_change(fill_method=None)
    historical_returns = historical_returns.replace([np.inf, -np.inf], np.nan)
    del close_panel, adjustment_panel, adjusted_close
else:
    historical_returns = None

if os.path.exists(GEO_MAP_PATH):
    geo_table = pd.read_csv(GEO_MAP_PATH, dtype={"stock": str})
    required_geo_columns = {"stock", "geo"}
    if not required_geo_columns.issubset(geo_table.columns):
        raise ValueError("stock_geo_map.csv must contain stock and geo columns.")
    geo_by_stock = geo_table.drop_duplicates("stock").set_index("stock")["geo"]
else:
    geo_by_stock = None
    if CONFIG["gamma_geo"] > 0:
        raise FileNotFoundError(
            "gamma_geo is positive but stock_geo_map.csv was not found."
        )

print("risk panel loaded:", historical_returns is not None)
print("geographic map loaded:", geo_by_stock is not None)
```

```python
class RegressionNN(nn.Module):
    def __init__(self, input_dim, hidden_dim=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1),
        )

    def forward(self, x):
        return self.net(x).squeeze(-1)


def differentiable_weights(signal):
    positive_signal = F.softplus(signal)
    return positive_signal / positive_signal.sum().clamp_min(1e-12)


def penalties_for_block(portfolio_signal, block):
    block_signal = portfolio_signal[block["row_positions"]]
    weights = differentiable_weights(block_signal)

    asset_loss = torch.mean(F.relu(weights - CONFIG["w_max"]) ** 2)

    geo_terms = []
    for group_positions, cap in block["geo_groups"]:
        exposure = weights[group_positions].sum()
        geo_terms.append(F.relu(exposure - cap) ** 2)
    geo_loss = (
        torch.stack(geo_terms).mean()
        if geo_terms
        else portfolio_signal.sum() * 0.0
    )

    if block["risk_returns"] is None:
        risk_loss = portfolio_signal.sum() * 0.0
        portfolio_variance = portfolio_signal.sum() * 0.0
    else:
        risk_weights = weights[block["risk_positions"]]
        risk_weights = risk_weights / risk_weights.sum().clamp_min(1e-12)
        # For demeaned historical return matrix R, mean((R @ w)^2)
        # equals w' Sigma w under the matching covariance convention.
        portfolio_return_history = block["risk_returns"] @ risk_weights
        portfolio_variance = torch.mean(portfolio_return_history ** 2)
        variance_limit = CONFIG["sigma_max"] ** 2
        risk_loss = F.relu(portfolio_variance - variance_limit) ** 2

    return asset_loss, geo_loss, risk_loss, portfolio_variance
```

```python
def choose_constraint_dates(date_values):
    unique_dates = np.sort(pd.unique(date_values))
    requested = CONFIG["constraint_dates"]
    if requested is None or requested >= len(unique_dates):
        return unique_dates
    positions = np.linspace(0, len(unique_dates) - 1, requested, dtype=int)
    return unique_dates[np.unique(positions)]


def build_constraint_blocks(fit_meta):
    blocks = []
    fit_dates = fit_meta["date"].to_numpy()

    for dt in choose_constraint_dates(fit_dates):
        row_positions = np.flatnonzero(fit_dates == dt)
        stocks = fit_meta.iloc[row_positions]["stock"].astype(str).to_numpy()

        geo_groups = []
        if geo_by_stock is not None and CONFIG["geo_caps"]:
            geo_labels = pd.Series(stocks).map(geo_by_stock).fillna("UNKNOWN")
            for geo_name, cap in CONFIG["geo_caps"].items():
                group_positions = np.flatnonzero(
                    geo_labels.to_numpy() == geo_name
                )
                if len(group_positions):
                    geo_groups.append(
                        (
                            torch.as_tensor(
                                group_positions, dtype=torch.long, device=device
                            ),
                            torch.tensor(float(cap), device=device),
                        )
                    )

        risk_positions = None
        risk_tensor = None
        if historical_returns is not None:
            end_position = historical_returns.index.searchsorted(
                pd.Timestamp(dt), side="left"
            )
            start_position = max(
                0, end_position - CONFIG["risk_lookback"]
            )
            history = historical_returns.iloc[start_position:end_position]
            history = history.reindex(columns=stocks)
            enough_history = (
                history.notna().sum(axis=0)
                >= CONFIG["minimum_risk_history"]
            ).to_numpy()

            if enough_history.sum() >= 2:
                risk_positions_np = np.flatnonzero(enough_history)
                history = history.iloc[:, risk_positions_np]
                history = history.fillna(history.mean()).fillna(0.0)
                history = history - history.mean(axis=0)
                risk_positions = torch.as_tensor(
                    risk_positions_np, dtype=torch.long, device=device
                )
                risk_tensor = torch.as_tensor(
                    history.to_numpy(dtype=np.float32), device=device
                )

        blocks.append(
            {
                "date": pd.Timestamp(dt),
                "row_positions": torch.as_tensor(
                    row_positions, dtype=torch.long, device=device
                ),
                "geo_groups": geo_groups,
                "risk_positions": risk_positions,
                "risk_returns": risk_tensor,
            }
        )

    return blocks


def train_constrained_mlp(
    model,
    X_fit,
    y_fit,
    primary_fit,
    fit_meta,
    X_validation,
    y_validation,
):
    scaler = StandardScaler()
    X_fit_scaled = scaler.fit_transform(X_fit)
    X_validation_scaled = scaler.transform(X_validation)

    X_fit_tensor = torch.as_tensor(
        X_fit_scaled, dtype=torch.float32, device=device
    )
    y_fit_tensor = torch.as_tensor(
        y_fit.to_numpy(), dtype=torch.float32, device=device
    )
    primary_fit_tensor = torch.as_tensor(
        primary_fit.to_numpy(), dtype=torch.float32, device=device
    )
    X_validation_tensor = torch.as_tensor(
        X_validation_scaled, dtype=torch.float32, device=device
    )
    y_validation_tensor = torch.as_tensor(
        y_validation.to_numpy(), dtype=torch.float32, device=device
    )

    model = model.to(device)
    optimizer = optim.Adam(
        model.parameters(),
        lr=CONFIG["learning_rate"],
        weight_decay=CONFIG["weight_decay"],
    )
    blocks = build_constraint_blocks(fit_meta.reset_index(drop=True))

    best_validation_loss = float("inf")
    best_state = None
    best_epoch = -1
    patience_counter = 0
    history_rows = []

    for epoch in range(CONFIG["epochs"]):
        model.train()
        optimizer.zero_grad()

        residual_prediction = model(X_fit_tensor)
        prediction_loss = F.mse_loss(residual_prediction, y_fit_tensor)

        # Constraints govern the pre-threshold final tradable signal.
        portfolio_signal = primary_fit_tensor + residual_prediction

        block_losses = [
            penalties_for_block(portfolio_signal, block) for block in blocks
        ]
        if block_losses:
            asset_loss = torch.stack([x[0] for x in block_losses]).mean()
            geo_loss = torch.stack([x[1] for x in block_losses]).mean()
            risk_loss = torch.stack([x[2] for x in block_losses]).mean()
            portfolio_variance = torch.stack([x[3] for x in block_losses]).mean()
        else:
            zero = prediction_loss * 0.0
            asset_loss = geo_loss = risk_loss = portfolio_variance = zero

        total_loss = (
            prediction_loss
            + CONFIG["gamma_asset"] * asset_loss
            + CONFIG["gamma_geo"] * geo_loss
            + CONFIG["gamma_risk"] * risk_loss
        )
        total_loss.backward()
        optimizer.step()

        model.eval()
        with torch.no_grad():
            validation_prediction = model(X_validation_tensor)
            validation_loss = F.mse_loss(
                validation_prediction, y_validation_tensor
            ).item()

        history_rows.append(
            {
                "epoch": epoch,
                "total_loss": total_loss.item(),
                "prediction_loss": prediction_loss.item(),
                "asset_loss": asset_loss.item(),
                "geo_loss": geo_loss.item(),
                "risk_loss": risk_loss.item(),
                "mean_portfolio_variance": portfolio_variance.item(),
                "validation_loss": validation_loss,
            }
        )

        if validation_loss < best_validation_loss:
            best_validation_loss = validation_loss
            best_epoch = epoch
            best_state = {
                key: value.detach().cpu().clone()
                for key, value in model.state_dict().items()
            }
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= CONFIG["patience"]:
            break

    if best_state is None:
        raise RuntimeError("Training ended without a valid model state.")
    model.load_state_dict(best_state)
    model.to(device)
    return model, scaler, pd.DataFrame(history_rows), best_epoch
```

```python
model_dates = np.sort(meta_features_model["date"].unique())
train_days = CONFIG["train_days"]
validation_days = CONFIG["validation_days"]
test_days = CONFIG["test_days"]
step_days = CONFIG["step_days"]
embargo_days = CONFIG["label_horizon_days"]

n_iterations = (
    (len(model_dates) - train_days - test_days) // step_days + 1
)
result_frames = []
diagnostic_rows = []

with tqdm(
    total=n_iterations,
    desc="Constrained walk-forward validation",
    ncols=100,
) as progress_bar:
    for iteration in range(n_iterations):
        set_seed(CONFIG["seed"] + iteration)
        start = iteration * step_days
        test_start = start + train_days
        test_end = test_start + test_days

        # Purged fitting/validation split for a forward-return label.
        validation_end = test_start - embargo_days
        validation_start = validation_end - validation_days
        fit_end = validation_start - embargo_days

        fit_dates = model_dates[start:fit_end]
        validation_dates = model_dates[validation_start:validation_end]
        test_dates = model_dates[test_start:test_end]

        fit_mask = meta_features_model["date"].isin(fit_dates)
        validation_mask = meta_features_model["date"].isin(validation_dates)
        test_mask = meta_features_model["date"].isin(test_dates)

        X_fit = meta_features_model.loc[fit_mask, feature_columns]
        y_fit = meta_targets.loc[fit_mask]
        primary_fit = primary_predictions.loc[fit_mask]
        fit_meta = meta_features_model.loc[fit_mask, ["date", "stock"]]

        X_validation = meta_features_model.loc[
            validation_mask, feature_columns
        ]
        y_validation = meta_targets.loc[validation_mask]

        X_test = meta_features_model.loc[test_mask, feature_columns]
        y_test = meta_targets.loc[test_mask]

        model = RegressionNN(
            input_dim=len(feature_columns),
            hidden_dim=CONFIG["hidden_dim"],
        )
        trained_model, scaler, training_history, best_epoch = (
            train_constrained_mlp(
                model,
                X_fit,
                y_fit,
                primary_fit,
                fit_meta,
                X_validation,
                y_validation,
            )
        )

        trained_model.eval()
        X_test_scaled = scaler.transform(X_test)
        X_test_tensor = torch.as_tensor(
            X_test_scaled, dtype=torch.float32, device=device
        )
        with torch.no_grad():
            residual_prediction = (
                trained_model(X_test_tensor).cpu().numpy()
            )

        primary_test = primary_predictions.loc[test_mask].to_numpy()
        true_test = true_returns.loc[test_mask].to_numpy()
        correction_is_active = (
            np.abs(residual_prediction) >= CONFIG["epsilon"]
        )
        filtered_correction = np.where(
            correction_is_active, residual_prediction, 0.0
        )
        final_signal = primary_test + filtered_correction

        residual_rmse = np.sqrt(
            mean_squared_error(y_test, residual_prediction)
        )
        residual_r2 = r2_score(y_test, residual_prediction)
        primary_rmse = np.sqrt(mean_squared_error(true_test, primary_test))
        final_rmse = np.sqrt(mean_squared_error(true_test, final_signal))

        test_keys = meta_features_model.loc[
            test_mask, ["date", "stock"]
        ].reset_index(drop=True)
        fold_result = test_keys.assign(
            true_return=true_test,
            primary_signal=primary_test,
            residual_true=y_test.to_numpy(),
            residual_prediction=residual_prediction,
            correction_active=correction_is_active,
            final_sig_with_constraints=final_signal,
            fold=iteration,
        )
        result_frames.append(fold_result)

        best_row = training_history.iloc[best_epoch]
        diagnostic_rows.append(
            {
                "fold": iteration,
                "test_start": pd.Timestamp(test_dates[0]),
                "test_end": pd.Timestamp(test_dates[-1]),
                "best_epoch": int(best_epoch),
                "validation_loss": best_row["validation_loss"],
                "prediction_loss": best_row["prediction_loss"],
                "asset_loss": best_row["asset_loss"],
                "geo_loss": best_row["geo_loss"],
                "risk_loss": best_row["risk_loss"],
                "mean_portfolio_variance": best_row[
                    "mean_portfolio_variance"
                ],
                "residual_rmse": residual_rmse,
                "residual_r2": residual_r2,
                "primary_rmse": primary_rmse,
                "final_rmse": final_rmse,
                "correction_activation_rate": correction_is_active.mean(),
            }
        )
        progress_bar.update(1)

constrained_results = pd.concat(result_frames, ignore_index=True)
constrained_diagnostics = pd.DataFrame(diagnostic_rows)
print(constrained_results.head())
print(constrained_diagnostics.tail())
```
